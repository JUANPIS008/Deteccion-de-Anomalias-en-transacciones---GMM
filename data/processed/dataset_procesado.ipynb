```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Cargar dataset original
url = 'https://raw.githubusercontent.com/JUANPIS008/datasets/main/credit_card_fraud.csv'
df = pd.read_csv(url)

# Calcular la cantidad total de valores nulos
total_nulos = df.isnull().sum().sum()

# Eliminar registros con valores faltantes si existen
if total_nulos > 0:
    df = df.dropna()


# Renombrar las variables originales para facilitar el procesamiento
df = df.rename(columns={
    'Transaction Amount': 'Amount',
    'Transaction Date and Time': 'Time'
})


# Convertir la fecha y hora a formato timestamp Unix en segundos
df['Time'] = pd.to_datetime(df['Time']).astype('int64') // 10**9


# Inicializar el escalador estándar
escalador = StandardScaler()

# Aplicar la transformación a las variables Amount y Time
df[['Amount', 'Time']] = escalador.fit_transform(df[['Amount', 'Time']])


# Mostrar todas las columnas disponibles
pd.set_option('display.max_columns', None)

# Mostrar dimensiones del dataset
print("Dimensiones del dataset procesado:")
print(df.shape)

# Mostrar los primeros registros
print("\nPrimeros 10 registros:")
display(df.head(10))

# Mostrar los últimos registros
print("\nÚltimos 10 registros:")
display(df.tail(10))

# Mostrar información general de la estructura del dataset
print("\nInformación general:")
df.info()

# Mostrar los tipos de datos de cada variable
print("\nTipos de datos:")
display(df.dtypes)

# Mostrar estadísticas descriptivas de las variables numéricas
print("\nEstadísticas descriptivas:")
display(df.describe())

# Mostrar los valores extremos de las variables transformadas
print("\nValores mínimos y máximos de las variables transformadas:")

print("\nTime")
print(f"Mínimo: {df['Time'].min():.4f}")
print(f"Máximo: {df['Time'].max():.4f}")

print("\nAmount")
print(f"Mínimo: {df['Amount'].min():.4f}")
print(f"Máximo: {df['Amount'].max():.4f}")

# Mostrar una muestra aleatoria del dataset procesado
print("\nMuestra aleatoria del dataset procesado:")
display(df.sample(10, random_state=42))
```
